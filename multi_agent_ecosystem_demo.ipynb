# 🌐 Multi-Model, Multi-Agent Collaborative Ecosystem

In this notebook, we demonstrate the **3-Agent Collaborative Network** with inter-agent communication, specialized domain tools, and secure file sharing:
- **Agent 01 (Product Query Assistant)**: Catalog DB, ChromaDB persistent memory & grounding, PDF invoices.
- **Agent 02 (Web Research Agent)**: Web reviews, competitor retail price tracking, market briefs.
- **Agent 03 (SQL & Data Analyst Agent)**: Safe SQL execution, category revenue metrics, visual bar charts.
- **Inter-Agent Bus & Secure Vault**: Agents delegating tasks to each other and securely exchanging reports.

## 1. Setup & Multi-Model Environment

```python
import sys
import os
from pathlib import Path
from dotenv import load_dotenv

# Add agent source paths
paths = [
    "src",
    "agents/01_product_query_agent/src",
    "agents/02_web_research_agent/src",
    "agents/03_data_analyst_agent/src",
    "shared"
]
for p in paths:
    if p not in sys.path:
        sys.path.insert(0, p)

load_dotenv("src/agentic_ai/.env")
load_dotenv(".env")

print("Environment initialized!")
print("GROQ_API_KEY Available:", bool(os.getenv("GROQ_API_KEY")))
```

## 2. Initialize the Multi-Agent Network Orchestrator

```python
from shared.orchestrator import MultiAgentNetwork

# Initialize network with multi-model provider
network = MultiAgentNetwork(provider="groq", model_name="qwen/qwen3.8-27b")

print("=== Registered Agents on Communication Bus ===")
for aid, desc in network.bus.get_registered_agents().items():
    print(f"🤖 {aid}: {desc}")
```

## 3. Testing Individual Domain Agents

### Agent 01: Product Catalog, ChromaDB Grounding & Specs

```python
res1 = network.agent_01.ask("What 4K monitors do you have in stock under $600 with USB-C charging?")
print(res1)
```

### Agent 02: Web Competitor Retail Price Intelligence

```python
res2 = network.agent_02.ask("Find all competitor retail prices for the Apple MacBook Air M3 and summarize the best deal.")
print(res2)
```

### Agent 03: SQL Database Analytics & Visual Revenue Chart

```python
res3 = network.agent_03.ask("Run SQL analytics to show net revenue by category and check which items have stockout risks within 14 days.")
print(res3)
```

## 4. End-to-End Multi-Agent Collaboration Workflow
All 3 agents cooperate sequentially and write an executive brief to the secure shared vault.

```python
collab_result = network.run_collaborative_workflow(
    goal="Evaluate the Dell UltraSharp 27 4K Monitor: catalog inventory & specs, external competitor prices, and sales performance."
)

print("=== Collaborative Workflow Complete! ===")
print(f"Saved to Shared Vault: {collab_result['shared_vault_report']}\n")
print(collab_result["final_summary"])
```

## 5. Inspecting the Inter-Agent Audit Log & Secure Shared Vault

```python
print("=== Shared Vault Files ===")
for f in network.vault.list_shared_files():
    print(f"📄 {f['filename']} ({f['size_bytes']} bytes)")

print("\n=== Inter-Agent Bus Communication Log ===")
for log in network.bus.get_audit_log():
    print(f"[{log['timestamp']}] {log['sender']} -> {log['recipient']} | Status: {log['status']}")
```
